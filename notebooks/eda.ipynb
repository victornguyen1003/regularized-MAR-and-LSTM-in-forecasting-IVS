# EDA

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/ivs_spx_16_26.csv')
display(df)
display(df.info())
display(df.describe())
```

```python
cols = df.columns.drop('Dates')
meta = pd.Series(cols).str.extract(r"(?P<tenor>\d+)(?P<unit>DAY|MTH)_IMPVOL_(?P<moneyness>[0-9.]+)%MNY_DF")
meta.head()
```

```python
tenor = np.sort(meta['tenor'].replace({'30': '1', '60': '2'}, regex=True).astype(int).unique())
moneyness = np.sort(meta['moneyness'].astype(float).unique())
print(f"Tenor: {tenor}")
print(f"Moneyness: {moneyness}")

cols_MultiIndex = pd.MultiIndex.from_product([tenor, moneyness], names=['tenor', 'moneyness'])

df_transformed = df.copy().set_index('Dates')
df_transformed.columns = cols_MultiIndex
df_transformed.head()
```

```python
date_to_plot = df['Dates'].iloc[0] 
print(f"Date: {date_to_plot}")

def display_and_plot(date):
    nx, ny = len(moneyness), len(tenor)
    X, Y = np.meshgrid(np.arange(nx), np.arange(ny))

    iv_matrix = df_transformed.loc[date].unstack(level='moneyness')
    
    display(iv_matrix)

    Z = iv_matrix.values

    fig, ax = plt.subplots(figsize=(100,8), subplot_kw={'projection': '3d'})
    ax.plot_surface(X, Y, Z, cmap="viridis")

    ax.set_xticks(np.arange(nx))
    ax.set_xticklabels(moneyness, fontsize=8, rotation=30)
    ax.set_xlabel('Moneyness (%)', fontsize=10, labelpad=10)

    ax.set_yticks(np.arange(ny))
    ax.set_yticklabels(tenor, fontsize=8)
    ax.set_ylabel("Tenor (M)", fontsize=10, labelpad=10)

    ax.set_title(f"IV surface of {date}", fontsize=12)

    plt.show()

try:
    display_and_plot(date_to_plot)
except KeyError:
    print(f"{date_to_plot} is not a trading date.")
```
